```python
import pandas as pd
from selenium.webdriver import Chrome
from selenium.webdriver.common.by import By
```

```python
driver = Chrome()
driver.maximize_window()
driver.get("https://www.w3resource.com/sql-exercises/sql-retrieve-from-table.php")
```

```python
article = driver.find_element(By.TAG_NAME, "article")
children = article.find_elements(By.XPATH, "./*")
```

```python
df = pd.DataFrame(columns=["q_text",])
idx, row = 0, 0
while idx < len(children):
    child = children[idx]
    txt: str = child.text
    if not txt:
        idx += 1
        continue
    txt = txt.split("\n")[0].strip()
    if txt[0].isdigit():
        if txt[1].isdigit():
            q_text = txt[3:].strip()
        else:
            q_text = txt[2:].strip()
        df.loc[row] = [q_text]
        row += 1
    idx += 1
```
